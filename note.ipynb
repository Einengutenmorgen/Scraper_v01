```python
import json

path = 'output/ria_analitika.jsonl'

max_entry = None
with open(path) as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        j = json.loads(line)
        if max_entry is None or j.get('paragraph_count', 0) > max_entry.get('paragraph_count', 0):
            max_entry = j

print(json.dumps(max_entry, ensure_ascii=False, indent=2))
```

```python
import json, re

path = 'output/ria_analitika.jsonl'

with open(path) as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        j = json.loads(line)
        if j['article_id'] == '2096556188':
            break

paras = [p for p in j['body'].split('\n') if p.strip()]
print("total blocks:", len(paras))

# categorize
bullets   = sum(1 for p in paras if p.lstrip().startswith('-'))
numbered  = sum(1 for p in paras if re.match(r'^\s*\d+[.)]', p))
headers   = sum(1 for p in paras if re.match(r'^\s*\d+\.\d', p))  # e.g. "1.1.", "2.4."
short     = sum(1 for p in paras if len(p) < 100)

print("bullet lines:  ", bullets)
print("numbered lines:", numbered)
print("sub-headers:   ", headers)
print("blocks <100ch: ", short)

# see the shortest blocks — these inflate the count
for p in sorted(paras, key=len)[:20]:
    print(repr(p[:80]))
    
```

```python
import json

path = "output/ria_analitika.jsonl"
recs = [json.loads(l) for l in open(path, encoding="utf-8") if l.strip()]
recs.sort(key=lambda r: r["word_count"])

# 5 evenly-spaced picks across the sorted-by-length set
picks = [recs[i] for i in (0, len(recs)//4, len(recs)//2, 3*len(recs)//4, -1)]

for r in picks:
    print("=" * 80)
    print(f"{r['article_id']} | {r['date']} | {r['byline']} "
          f"({'signed' if r['has_byline'] else 'inst'})")
    print(f"words={r['word_count']}  paras={r['paragraph_count']} "
          f"(prose {r['prose_paragraph_count']})  sents={r['sentence_count']}")
    print(f"TITLE: {r['title']}")
    print(f"{r['url']}")
    print("-" * 80)
    print(r["body"][:600] + (" …" if len(r["body"]) > 600 else ""))
    print()
    
```
